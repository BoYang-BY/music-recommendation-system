```python
import pandas as pd

# 讀取 Spotify dataset
df = pd.read_csv(
    "../dataset/spotify_track_dataset.csv"
)

# 移除沒有意義的索引欄
df.drop(
    columns=["Unnamed: 0"],
    inplace=True
)

# 查看資料
df.head()
```

```python
ncols, nrows = df.shape
print(f'Dataset has {ncols} rows and {nrows} columes')
```

```python
# 檢查重複資料數量
duplicated_rows = df.duplicated().sum()

if duplicated_rows == 0:
    print("資料中沒有重複的列，因此不需要進一步處理重複資料。")
else:
    print(f"資料中有 {duplicated_rows} 列是重複的，將這些資料刪除。")

    df = df.drop_duplicates()

    print(
        f"刪除重複列後，剩下 {df.shape[0]} 列資料。"
    )
```

```python
# =========================
# 處理缺失值 (Null)
# =========================

# 計算每一列(row)有多少個缺失值
missing_values_per_row = df.isnull().sum(axis=1)

# 統計：
# 有0個缺失值的列有多少
# 有1個缺失值的列有多少...
count_per_missing_value = (
    missing_values_per_row
    .value_counts()
    .sort_index()
)

# 輸出每種缺失值數量對應的列數
for missing, rows in count_per_missing_value.items():
    print(
        f"{rows} 列有 {missing} 個缺失值"
    )


# 計算至少有一個缺失值的資料列數
total_rows_with_missing_values = (
    df.isnull()
    .any(axis=1)
    .sum()
)

print(
    f"總共有 {total_rows_with_missing_values} 列含有缺失值"
)


# 找出哪些欄位含有缺失值
features_with_nan = [
    feature
    for feature in df.columns
    if df[feature].isna().sum() > 0
]


print(
    f"含有缺失值的欄位：{features_with_nan}"
)


# 刪除所有含缺失值的資料列
df = df.dropna()


print(
    f"已刪除含缺失值的列，更新後 DataFrame 形狀為：{df.shape}"
)


# 再次確認目前沒有缺失值
df.isna().sum()
```

```python
# =========================
# 檢查 time_signature 範圍
# =========================

# 計算 time_signature 小於 3 的數量與比例
num_below_3 = len(
    df[df["time_signature"] < 3]
)

pct_below_3 = (
    num_below_3 / len(df) * 100
)

print(
    f"{pct_below_3:.2f}% 的資料其 time_signature 小於 3"
)


# 計算 time_signature 大於 7 的數量與比例
num_above_7 = len(
    df[df["time_signature"] > 7]
)

pct_above_7 = (
    num_above_7 / len(df) * 100
)

print(
    f"{pct_above_7:.2f}% 的資料其 time_signature 大於 7"
)


# 刪除 time_signature 小於 3 的異常資料
print(
    "刪除前，資料形狀為：",
    df.shape
)

df = df.loc[
    df["time_signature"] >= 3
]


print(
    "刪除後，資料形狀為：",
    df.shape
)
```

```python
# =========================
# 檢查 duration_ms 是否為 0
# =========================

print(
    "刪除 duration_ms 為 0 前，資料形狀為：",
    df.shape
)


df = df[
    df["duration_ms"] != 0
]


print(
    "刪除 duration_ms 為 0 後，資料形狀為：",
    df.shape
)
```

```python
import numpy as np

# 儲存欄位名稱
numeric_cols = []
categorical_cols = []


# 判斷欄位類型
for col in df.columns:

    # 數值型欄位
    if (
        df[col].dtype == np.float64
        or df[col].dtype == np.int64
    ):
        numeric_cols.append(col)

    # 其他視為類別型
    else:
        categorical_cols.append(col)



# 找出離散數值型欄位
# 唯一值少於20，之後當類別處理

feature_discrete_numerical = [
    feature
    for feature in numeric_cols
    if df[feature].nunique() < 20
]


# 從連續數值中移除離散數值
numeric_cols = [
    feature
    for feature in numeric_cols
    if feature not in feature_discrete_numerical
]


print("連續數值型欄位：")
print(numeric_cols)


print("\n非數值型類別欄位：")
print(categorical_cols)


print("\n離散數值型欄位：")
print(feature_discrete_numerical)
```

```python
df["time_signature"] = df["time_signature"].replace(
    {
        3: "3/4",
        4: "4/4",
        5: "5/4",
        6: "6/4",
        7: "7/4"
    }
)
```

```python
df["key"] = df["key"].replace(
    {
        0: "C",
        1: "C-sharp_D-flat",
        2: "D",
        3: "D-sharp_E-flat",
        4: "E",
        5: "F",
        6: "F-sharp_G-flat",
        7: "G",
        8: "G-sharp_A-flat",
        9: "A",
        10: "A-sharp_B-flat",
        11: "B"
    }
)
```

```python
df["mode"] = df["mode"].replace(
    {
        0: "minor",
        1: "major"
    }
)

df["explicit"] = np.where(
    df["explicit"],
    "yes",
    "no"
)
```

```python
speechiness_type = []


for value in df["speechiness"]:

    if value < 0.33:
        speechiness_type.append("Low")

    elif value <= 0.66:
        speechiness_type.append("Medium")

    else:
        speechiness_type.append("High")


# 新增欄位
df["speechiness_type"] = speechiness_type


print(
    df["speechiness_type"].value_counts()
)
```

```python
# =========================
# 分割資料
# =========================

cutoff = 100000


# 前 100000 筆資料
subset1_df = df.iloc[:cutoff, :]


# 剩餘資料
subset2_df = df.iloc[cutoff:, :]


print(subset1_df.shape)
print(subset2_df.shape)



# 儲存兩份資料
subset1_df.to_csv(
    "../dataset/spotify_track_dataset_current.csv",
    index=False
)


subset2_df.to_csv(
    "../dataset/spotify_track_dataset_new.csv",
    index=False
)
```

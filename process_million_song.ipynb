```python
import tables
```

```python
def explore_h5_file(h5filename):
    # 開啟 HDF5 檔案
    h5file = hdf5_getters.open_h5_file_read(h5filename)

    print(f"File structure of: {h5filename}\n")

    # 走訪所有節點
    for node in h5file.walk_nodes():
        print(f"Node: {node._v_pathname}, Type: {type(node)}")

        # 如果是 Table，就印出內容
        if isinstance(node, tables.Table):
            print(f"Dataset Name: {node.name}")

            try:
                print(node.read())
            except Exception as e:
                print(f"Could not read {node.name}: {e}")

            print("-" * 80)

    h5file.close()
```

```python
explore_h5_file(file_path)
```
